## 1. Programación Orientada a Objetos POO

En programación orientada a objetos (POO), una clase es un elemento que define las características y comportamientos de una entidad, como un molde para crear objetos.
La POO es un paradigma de programación que se basa en el concepto de objetos, que son entidades que combinan datos (atributos) y comportamientos (métodos). Las clases actúan como plantillas para crear objetos, definiendo su estructura y comportamiento.
Las clases son un concepto fundamental en la POO y se utilizan para modelar entidades del mundo real o abstracto en un programa de computadora.


##  ¿Qué es una Clase?
* **Plantilla o molde**: Una clase es como una plantilla que define las características y comportamientos de un conjunto de objetos.

* **Atributos**: Son las propiedades o características que describen a un objeto (por ejemplo, el color de un auto, la edad de una persona).

* **Métodos**: Son las acciones o funciones que un objeto puede realizar (por ejemplo, arrancar un auto, hablar, correr).

```python
#Ejemplo clase perro
class Perro:
    def __init__(self, nombre, raza):
        self.nombre = nombre
        self.raza = raza

    def ladrar(self):
        print(f"{self.nombre} dice: Guau!")
```

**En este ejemplo**:

* `Perro` es la clase.
* `nombre` y `raza` son atributos.
* `ladrar` es un método.

**¿Qué es un Objeto?**
* Instancia de una clase: Un objeto es una instancia concreta de una clase.
* Tiene sus propios valores: Cada objeto tiene sus propios valores para los atributos definidos en la clase.

```python
# Instanciar dos objetos de la clase perro
mi_perro = Perro("Firulais", "Labrador")
otro_perro = Perro("Snoopy","Beagle")
```

```python
# mi_perro llama al método ladrar
mi_perro.ladrar()
otro_perro.ladrar()
```

## Principios Fundamentales de la POO

* **Abstracción**: Simplifica la realidad al enfocarse en los detalles relevantes de un objeto y ocultar lo que no es esencial. Por ejemplo, en una clase Coche, sólo incluimos atributos y métodos como color, marca y arrancar(), omitiendo detalles como el cableado interno.

* **Encapsulamiento**: Protege los datos de un objeto restringiendo el acceso directo a sus atributos. Solo se accede a los datos a través de métodos específicos (getter/setter). Esto permite controlar cómo se modifican y acceden los datos, mejorando la seguridad y flexibilidad.

* **Herencia**: Permite crear nuevas clases basadas en clases existentes, reutilizando su código y añadiendo o modificando funcionalidades. Esto ayuda a evitar duplicación de código y facilita la escalabilidad. Por ejemplo, una clase Perro podría heredar de una clase Animal.

* **Polimorfismo**: Permite usar una misma interfaz para diferentes tipos de objetos. Esto significa que puedes llamar al mismo método en objetos de distintas clases, y cada objeto puede comportarse de manera diferente. Así, puedes tener un método hablar() para Perro y Gato, que producen sonidos distintos.

## 3. Atributos y Metodos
### 3.1 Atributos
Todas las variables o atributos de la clase, se guardan dentro de la clase por medio del objeto self (en español "uno mismo"). Tambien se pueden definir los atributos inmediatamente despues de definir el nombre de la clase y su tipo.

```python
class Rectangulo:
    # Definimos el constructor con atributos públicos
    def __init__(self, ancho=0, alto=0):
        self.ancho = ancho
        self.alto = alto

    # Método para calcular el área
    def calcular_area(self):
        return self.ancho * self.alto

    # Método para calcular el perímetro
    def calcular_perimetro(self):
        return 2 * (self.ancho + self.alto)
```

```python
# Uso de la clase
rectangulo = Rectangulo()

# Accedemos a los atributos directamente (son públicos)
print(f"Ancho del rectángulo: {rectangulo.ancho}")
print(f"Alto del rectángulo: {rectangulo.alto}")

# Calculamos el área y el perímetro usando los métodos de la clase
print(f"Área del rectángulo: {rectangulo.calcular_area()}")
print(f"Perímetro del rectángulo: {rectangulo.calcular_perimetro()}")
```

# Explicación del Ejemplo
* **Atributos Públicos**: Los atributos `ancho` y `alto` son públicos, ya que son datos que cualquier usuario de la clase puede ver y modificar directamente. En este caso, estos valores representan propiedades fundamentales del rectángulo, por lo que no es necesario ocultarlos.

* **Métodos de Abstracción**: Los métodos `calcular_area` y `calcular_perimetro` proporcionan una abstracción del cálculo sin que el usuario necesite preocuparse por las fórmulas matemáticas detrás de ellos. Solo necesita llamar a estos métodos para obtener el área o el perímetro.

* Ocultación de Lógica de Cálculo: Aunque `ancho` y `alto` son atributos públicos, la lógica de cómo se calcula el área y el perímetro está *encapsulada* dentro de los métodos de la clase. Esto permite que cualquier cambio en la forma de cálculo solo afecte al código dentro de la clase y no al código que utiliza esta clase.

En Python, cuando los atributos de una clase están definidos como privados, generalmente se les coloca un guion bajo `_` o doble guion bajo `__` al inicio del nombre, lo que sugiere que no deben accederse directamente desde fuera de la clase. Para manipular estos atributos, es común implementar métodos *getter* y *setter* que permiten leer y modificar los valores de manera controlada.

Ejemplo: Implementar la clase persona, con atributos nombre, edad

```python
class Persona:
    def __init__(self, nombre, edad):
        self.__nombre = nombre  # Atributo privado
        self.__edad = edad      # Atributo privado

    # Método getter para el atributo nombre
    def get_nombre(self):
        return self.__nombre

    # Método setter para el atributo nombre
    def set_nombre(self, nombre):
        self.__nombre = nombre

    # Método getter para el atributo edad
    def get_edad(self):
        return self.__edad

    # Método setter para el atributo edad
    def set_edad(self, edad):
        if edad >= 0:  # Agregamos una validación
            self.__edad = edad
        else:
            print("La edad no puede ser negativa")

# Uso de los métodos getter y setter
persona = Persona("Juan", 25)
print(persona.get_nombre())  # Accede al nombre usando el getter
persona.set_nombre("Carlos")  # Modifica el nombre usando el setter
print(persona.get_nombre())

print(persona.get_edad())  # Accede a la edad usando el getter
persona.set_edad(30)  # Modifica la edad usando el setter
print(persona.get_edad())
```

## Explicación
* **Definición de Atributos Privados**: Los atributos `__nombre` y `__edad` están precedidos por dos guiones bajos, lo que hace que Python los trate como privados.

* **Métodos Getter**: `get_nombre` y `get_edad` permiten acceder a los valores de `__nombre` y `__edad` respectivamente, sin hacerlo directamente.

* **Métodos Setter**: `set_nombre` y `set_edad` permiten modificar los valores de `__nombre` y `__edad`. En `set_edad`, se agrega una verificación para asegurar que la edad no sea negativa, lo cual es un ejemplo de cómo los setters pueden controlar la modificación del valor.

1) Ejemplo de abstracción mediante una clase `Cafetera`. En este caso, creamos una cafetera que permite hacer café sin exponer detalles de implementación internos (como la cantidad de agua o el nivel de café).


```python
class Cafetera:
    def __init__(self):
        # Estos son atributos "privados" que no son accesibles directamente
        self._agua = 1000  # ml
        self._cafe = 500   # gramos

    def preparar_cafe(self):
        if self._agua >= 200 and self._cafe >= 15:
            self._agua -= 200
            self._cafe -= 15
            print("Café preparado!")
        else:
            print("No hay suficiente agua o café para preparar la bebida.")

    def llenar_agua(self):
        self._agua = 1000
        print("Tanque de agua lleno.")

    def llenar_cafe(self):
        self._cafe = 500
        print("Café en grano lleno.")
```

```python
# Uso de la clase
mi_cafetera = Cafetera()
mi_cafetera.preparar_cafe()    # Produce café si hay suficiente agua y café
mi_cafetera.preparar_cafe()    # Prepara otra taza
mi_cafetera.llenar_agua()      # Llena el tanque de agua
mi_cafetera.llenar_cafe()      # Llena el depósito de café
```

### Explicación del Ejemplo
* **Atributos Privados**: Los atributos `_agua` y `_cafe` están precedidos por un guion bajo (`_`), lo que indica que son detalles internos de la clase que no deben ser manipulados directamente fuera de ella. Esto representa el *encapsulamiento* y la *abstracción*, ya que no necesita conocer ni manipular la cantidad de agua y café para usar la cafetera.

* **Métodos Públicos**: Los métodos `preparar_cafe`, `llenar_agua` y `llenar_cafe` son los únicos métodos disponibles para interactuar con la cafetera. Estos métodos abstraen los detalles de cómo se gestionan el agua y el café dentro de la máquina, permitiendo solo las operaciones necesarias para el usuario.

* Ocultación de Detalles: El usuario no necesita saber cuántos mililitros de agua o gramos de café están en la máquina; solo llama a `preparar_cafe()` y la cafetera se encarga del resto.

##¿Por Qué se Utiliza la POO?
1. **Modularidad y Reusabilidad**: La POO facilita dividir el código en partes autónomas y reutilizables. Las clases pueden ser usadas en múltiples partes de un programa o incluso en diferentes proyectos.

2. **Mantenimiento y Escalabilidad**: El código orientado a objetos es más fácil de mantener y escalar, ya que los cambios en una clase afectan a todos los objetos derivados de ella. Además, la herencia y el polimorfismo permiten modificar y extender el código sin alterar el funcionamiento básico del programa.

3. **Facilidad para Modelar Problemas Complejos**: La POO permite modelar sistemas complejos de una manera organizada y lógica, ya que los objetos representan componentes o conceptos del problema que se está resolviendo. Esto facilita la comprensión del código y la resolución de problemas.

4. **Abstracción y Ocultamiento de Información**: Al usar encapsulamiento, se logra ocultar la implementación interna de una clase, exponiendo solo los métodos necesarios. Esto reduce la probabilidad de errores y mejora la seguridad del código.

5. **Simulación de Entornos Reales**: La POO es especialmente útil en aplicaciones donde se simulan sistemas reales, como videojuegos, simuladores de entrenamiento o software de ingeniería. Permite crear objetos que se comportan de manera similar a las entidades que representan en la vida real.

### 3.2. Metodos
Los metodos son las funciones que se definen dentro de la clase. Los metodos pueden emplear atributos o argumentos externos al momento de ser invocados. Por ejemplo creamos una clase llamada `vector2d` que almacena las componentes en los ejes horizontal y vertical.
```python
class vector2d:
  def __init__(self, x, y):
    self.x = x
    self.y = y
  def suma(self, vector):
    # Sumar dos Vectores
    return vector2d(self.x + vector.x, self.y + vector.y)
```

#Operadores comunes para sobrecargar
##Aritméticos:

* `+ (__add__):` Suma.
* `- (__sub__):` Resta.
* `(__mul__):` Multiplicación.
* `/ (__truediv__)`: División.
* `// (__floordiv__)`: División entera.

##Comparación:

* `== (__eq__):` Igualdad.
* `!= (__ne__):` Desigualdad.
* `< (__lt__):` Menor que.
* `<= (__le__):` Menor o igual.
* `> (__gt__):` Mayor que.
* `>= (__ge__)`: Mayor o igual.

##Representación:

* `str(obj) (__str__):` Representación amigable.
* `repr(obj) (__repr__):` Representación técnica.

##Acceso y manipulación:

* `[] (__getitem__, __setitem__)`: Acceso a índices.


```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        """Suma de dos vectores."""
        if isinstance(other, Vector2D):
            return Vector2D(self.x + other.x, self.y + other.y)
        return NotImplemented

    def __sub__(self, other):
        """Resta de dos vectores."""
        if isinstance(other, Vector2D):
            return Vector2D(self.x - other.x, self.y - other.y)
        return NotImplemented

    def __mul__(self, scalar):
        """Multiplicación por un escalar."""
        if isinstance(scalar, (int, float)):
            return Vector2D(self.x * scalar, self.y * scalar)
        return NotImplemented

    def __eq__(self, other):
        """Comparación de igualdad entre vectores."""
        if isinstance(other, Vector2D):
            return self.x == other.x and self.y == other.y
        return NotImplemented

    def __repr__(self):
        """Representación técnica del vector."""
        return f"Vector2D(x={self.x}, y={self.y})"

    def __str__(self):
        """Representación amigable del vector."""
        return f"({self.x}, {self.y})"

# Ejemplo de uso
v1 = Vector2D(2, 3)
v2 = Vector2D(1, 4)

# Suma
v3 = v1 + v2  # Vector2D(x=3, y=7)
print("Suma:", v3)

# Resta
v4 = v1 - v2  # Vector2D(x=1, y=-1)
print("Resta:", v4)

# Multiplicación por escalar
v5 = v1 * 3  # Vector2D(x=6, y=9)
print("Multiplicación por escalar:", v5)

# Comparación
print("¿Son iguales?", v1 == v2)  # False
```

## Ejercicio 1
- Diseñar una clase para modelar vectores de n-dimensiones, por medio de listas. Debe tener como atributos: los valores y el tamaño del vector. Sobrecargar la clase de tal manera que se puede realizar la suma entre vectores elemento a elemento. Tambien sobrecargar la resta y la multiplicacion.

```python
class Vector:
    def __init__(self, components):
        if not isinstance(components, list) or not all(isinstance(c, (int, float)) for c in components):
            raise ValueError("Los componentes deben ser una lista de números.")
        self.components = components

    def __add__(self, other):
        """Suma de dos vectores."""
        if isinstance(other, Vector) and len(self.components) == len(other.components):
            return Vector([a + b for a, b in zip(self.components, other.components)])
        raise ValueError("Ambos vectores deben tener la misma dimensión.")

    def __sub__(self, other):
        """Resta de dos vectores."""
        if isinstance(other, Vector) and len(self.components) == len(other.components):
            return Vector([a - b for a, b in zip(self.components, other.components)])
        raise ValueError("Ambos vectores deben tener la misma dimensión.")

    def __mul__(self, scalar):
        """Multiplicación por un escalar."""
        if isinstance(scalar, (int, float)):
            return Vector([c * scalar for c in self.components])
        raise ValueError("El escalar debe ser un número.")

    def __eq__(self, other):
        """Comparación de igualdad entre vectores."""
        if isinstance(other, Vector):
            return self.components == other.components
        return False

    def __len__(self):
        """Número de dimensiones del vector."""
        return len(self.components)

    def __repr__(self):
        """Representación técnica del vector."""
        return f"Vector({self.components})"

    def __str__(self):
        """Representación amigable del vector."""
        return f"Vector: {self.components}"




```

```python
# Ejemplo de uso
v1 = Vector([1, 2, 3])
v2 = Vector([4, 5, 6])

# Suma
v3 = v1 + v2  # Vector([5, 7, 9])
print("Suma:", v3)

# Resta
v4 = v1 - v2  # Vector([-3, -3, -3])
print("Resta:", v4)

# Multiplicación por escalar
v5 = v1 * 2  # Vector([2, 4, 6])
print("Multiplicación por escalar:", v5)

# Comparación
print("¿Son iguales?", v1 == v2)  # False

# Número de dimensiones
print("Dimensiones de v1:", len(v1))  # 3
```

## 5. Herencia
Es un concepto fundamental en la programacion orientada a objetos. La idea es crear clases u objetos a partir de otra clase, y heredar los atributos y metodos de la clase madre/padre. Como ejemplo creamos una clase animal y una subclase Perro:

```python
class Animal:
  def __init__(self,nombre):
    self.nombre = nombre

  def hacer_sonido(self):
    print("el animal hace un sonido!!")

  def caminar(self):
    print("el animal camina en cuatro patas")

```

```python
class Perro(Animal):
  def hacer_sonido(self):
    print("Guau, Guau!!")
```

```python
A1 = Animal("Firulais")
A1.hacer_sonido()
```

```python
P1 = Perro("Snoopy")
P1.hacer_sonido()
P1.caminar()
```

##Características principales de la herencia

* **Reutilización de código**: Las subclases no necesitan reescribir métodos o atributos ya definidos en la superclase.

* **Especialización**: Las subclases pueden agregar o sobrescribir (modificar) métodos y atributos de la superclase para adaptarlos a necesidades específicas.

* **Jerarquía**: Crea una estructura jerárquica entre clases, modelando relaciones "es-un" (por ejemplo, "un gato es-un animal").

* **Polimorfismo**: Las subclases pueden redefinir métodos de la superclase, permitiendo comportamientos específicos según el contexto.

##Ejemplo práctico: Banco
Un banco tiene cuentas básicas y cuentas premium:




```python
class CuentaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, cantidad):
        self.saldo += cantidad
        return f"Saldo actualizado: ${self.saldo}"

    def retirar(self, cantidad):
        if cantidad > self.saldo:
            return "Fondos insuficientes."
        self.saldo -= cantidad
        return f"Saldo actualizado: ${self.saldo}"

# Subclase
class CuentaPremium(CuentaBancaria):
    def __init__(self, titular, saldo=0, beneficios=[]):
        super().__init__(titular, saldo)  # Reutilizamos el constructor de la superclase
        self.beneficios = beneficios

    def mostrar_beneficios(self):
        return f"Beneficios premium: {', '.join(self.beneficios)}"

# Uso
cuenta = CuentaPremium("Juan", saldo=1000, beneficios=["Cashback", "Tasa baja"])
print(cuenta.depositar(500))            # Saldo actualizado: $1500
print(cuenta.retirar(300))              # Saldo actualizado: $1200
print(cuenta.mostrar_beneficios())      # Beneficios premium: Cashback, Tasa baja
```

### 5.3 Herencia multiple


La herencia múltiple es un concepto en la programación orientada a objetos donde una clase (llamada clase hija o subclase) puede heredar atributos y métodos de más de una clase base o superclase. Esto permite combinar funcionalidades de diferentes clases en una sola.

Python admite herencia múltiple, lo que lo hace flexible, pero requiere precaución para evitar conflictos y complicaciones en el diseño del código.

##Sintaxis

```python
class ClasePadre1:
    # Métodos y atributos de la primera superclase
    pass

class ClasePadre2:
    # Métodos y atributos de la segunda superclase
    pass

class ClaseHija(ClasePadre1, ClasePadre2):
    # La clase hija hereda de ambas superclases
    pass
```

```python

#Supongamos que queremos modelar animales que vuelan y animales acuáticos:
# Primera superclase
class Volador:
    def volar(self):
        return "Estoy volando."

# Segunda superclase
class Acuatico:
    def nadar(self):
        return "Estoy nadando."

# Clase hija
class Pato(Volador, Acuatico):
    def caminar(self):
        return "Estoy caminando."

# Uso
pato = Pato()
print(pato.volar())   # Estoy volando.
print(pato.nadar())   # Estoy nadando.
print(pato.caminar()) # Estoy caminando.
```

```python
# Definicion clase para Padre
class Padre():
	# Metodo de la clase Padre
	def canta(self):
		print("Papá canta bien.")

# Definicion clase para Madre
class Madre():
	# Metodo de la clase Madre
	def programa(self):
		print("Mamá programa bien.")

# Definicion de la clase derivada
class Hijo(Padre, Madre):
	def jugar(self):
		print("Al niño le encanta jugar")

# creando un objeto de la clase derivada
niño = Hijo()
# llamando a los metodos de las clases padres y la derivada
niño.canta()
niño.programa()
niño.jugar()
```

crear una clase calculadora cuyo atributo sea una lista con el historial de operaciones. Los métodos son las operaciones así:

* sumar(a,b)
* restar(a,b)
* multiplicar(a,b)
* dividir(a,b)
* mostrar_historial()

* Calc.sumar(1,2)
* Calc.restar(3,4)

["1 + 2 = 3", "3 - 4 = -1"]

```python
class calculadora:
  def __init__(self,historial = dict()):
    self.historial = historial


  def sumar(self,a,b):
    if "SUMA" in self.historial:
      self.historial["SUMA"] +=", " + str(a) + " + " + str(b) +" = " + str(a+b)
    else:
      self.historial["SUMA"] = str(a) + " + " + str(b) +" = " + str(a+b)

  def restar(self,a,b):
    if "RESTA" in self.historial:
      self.historial["RESTA"] +=", " + str(a) + " - " + str(b) +" = " + str(a-b)
    else:
      self.historial["RESTA"] = str(a) + " - " + str(b) +" = " + str(a-b)




  def mostrar_historial(self):
    print(self.historial)
```

```python
Calc = calculadora()
Calc.sumar(2,3)
Calc.sumar(15,9)
Calc.sumar(8,43)
Calc.restar(7,21)
Calc.restar(43,198)
Calc.mostrar_historial()
H = Calc.historial
```

```python
Calc2 = calculadora(H)
Calc2.sumar(9,8)
Calc2.mostrar_historial()
```

clase que se denomine producto, cuyos atributos sean nombre, precio, cantidad, metodo de representacion

crear una clase inventario cuyo atributo sea una lista de productos y que tenga los metods agregar_producto,  eliminar_producto, mostrar inventario

```python
class producto:
  def __init__(self,nombre = "",precio= 0,cantidad = 0):
    self.nombre = nombre
    self.precio = precio
    self.cantidad = cantidad

  def __str__(self):
    return f"{self.nombre}, valor = ${self.precio}, cantidad = {self.cantidad}"
```

```python
class inventario:
  def __init__(self,productos = list()):
    self.productos = productos

  def agregar_producto(self,producto):
    self.productos.append(producto)

  def eliminar_producto(self,nombre):
    ind = 0
    flag = False

    for i in self.productos:
      if i.nombre == nombre:
        j = ind
        flag = True
      ind += 1
    if flag:
      self.productos.pop(j)
    else:
      print(f"el producto {nombre} no se encuentra en el inventario")


  def mostrar_inventario(self):
    for i in self.productos:
      print(i)
```

```python
P1 = producto("Aguacates",3500,3)
P2 = producto("Tomates",500,15)
P3 = producto("Bananos",300,30)

Inv1 = inventario([P1,P2])
Inv1.agregar_producto(P3)
Inv1.mostrar_inventario()
Inv1.agregar_producto(producto("Papas",800,78))
Inv1.eliminar_producto("Tomates")
print(" ")
Inv1.mostrar_inventario()
```
